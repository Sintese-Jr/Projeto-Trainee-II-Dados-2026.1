## Importe de bibliotecas

```python
import pandas as pd
import joblib
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report
from sklearn.preprocessing import StandardScaler
```

## Formatação do Dataset teste

```python
df_conv = pd.read_csv('../dados/teste/dados_teste.csv')


colunas_numericas = ['dias_desde_ultimo_acesso', 'series_concluidas', 'series_abandonadas', 'tickets_suporte_abertos']
scaler = StandardScaler()
scaler.fit(df_conv[colunas_numericas])

df_conv = df_conv.rename(columns={
    'camcelou': 'cancelou',
    'metodo_de_pagamento': 'pagamento',
    'plano_atual': 'plano'
})
df_conv['genero_favorito'] = df_conv['genero_favorito'].fillna('Desconhecido')
df_conv = df_conv.dropna(subset=['plano'])
df_conv = df_conv.drop(columns=['id', 'nome_do_cliente', 'data_de_assinatura', 'idade_do_usuario', 'idade_do_usuário', 'horas_consumidas', 'plano', 'pagamento', 'dispositivo_mais_acesso'], errors='ignore')
df_conv[colunas_numericas] = scaler.transform(df_conv[colunas_numericas])
df_conv = pd.get_dummies(df_conv, columns=['genero_favorito'], drop_first=False, dtype=int)
```

## Teste e Métricas

```python
# Carregamento dos Dados
caminho_modelo = "../modelo/modelo.pkl"
caminho_entrada = df_conv
coluna_real = "cancelou"

# Carregamento de Modelo
modelo = joblib.load(caminho_modelo)
df_entrada = caminho_entrada

X_teste = df_entrada.drop(columns=[coluna_real])
y_real = df_entrada[coluna_real]
y_pred = modelo.predict(X_teste)

acuracia = accuracy_score(y_real, y_pred)
matriz_conf = confusion_matrix(y_real, y_pred)

print("=" * 40)
print(f"Acurácia do Modelo: {acuracia * 100:.2f}%")
print("=" * 40)

print("\nMatriz de Confusão:")
print(matriz_conf)

print("\nRelatório Completo (Precision, Recall, F1-Score):")
print(classification_report(y_real, y_pred))
```
